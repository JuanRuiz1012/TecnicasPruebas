```python
import os
import pandas as pd

# Directorio actual
folder_path = '.'

# Obtener lista de archivos CSV y XLSX
files = [f for f in os.listdir(folder_path) if f.endswith('.csv') or f.endswith('.xlsx')]

print(f"=== SE ENCONTRARON {len(files)} ARCHIVOS PARA ANALIZAR ===\n")

for file_name in files:
    full_path = os.path.join(folder_path, file_name)
    size_mb = os.path.getsize(full_path) / (1024 * 1024)
    print("=" * 80)
    print(f"📄 ARCHIVO: {file_name} ({size_mb:.2f} MB)")
    print("=" * 80)
    
    try:
        # Cargar una muestra reducida (50 filas) para no saturar la memoria RAM en la inspección
        if file_name.endswith('.csv'):
            # Probar con encoding utf-8 o latin-1 por si hay tildes/caracteres de Colombia
            try:
                df_sample = pd.read_csv(full_path, nrows=50, low_memory=False)
            except UnicodeDecodeError:
                df_sample = pd.read_csv(full_path, nrows=50, encoding='latin-1', low_memory=False)
        else:
            df_sample = pd.read_excel(full_path, nrows=50)
            
        print(f"📌 Total de columnas: {len(df_sample.columns)}")
        print("📌 Lista de columnas:")
        for idx, col in enumerate(df_sample.columns, 1):
            print(f"   {idx}. {col}")
            
        print("\n🔍 Primeras 2 filas de muestra:")
        display(df_sample.head(2))
        
    except Exception as e:
        print(f"❌ Error al leer el archivo {file_name}: {e}")
        
    print("\n")
```

```python
import pandas as pd
import numpy as np
import os

print("⌛ Iniciando el proceso ETL de consolidación ajustado...")

dfs_procesados = []

def normalizar_gravedad(val):
    if pd.isna(val): return "OTRO"
    val = str(val).upper()
    if "HERID" in val: return "CON HERIDOS"
    if "MUERT" in val or "FATAL" in val: return "CON MUERTOS"
    if "DAÑO" in val or "SOLO DAÑO" in val: return "SOLO DAÑOS"
    return "OTRO"

def normalizar_clase(val):
    if pd.isna(val): return "OTRO"
    val = str(val).upper()
    if "CHOQUE" in val: return "CHOQUE"
    if "ATROPELLO" in val: return "ATROPELLO"
    if "VOLCAMIENTO" in val: return "VOLCAMIENTO"
    if "CAIDA" in val or "CAÍDA" in val: return "CAIDA OCUPANTE"
    return "OTRO"

# 1. BUCARAMANGA
try:
    print("🔄 Procesando Bucaramanga...")
    df_buca = pd.read_csv('3._Accidentes_de_Transito_ocurridos_en_el_Municipio_de_Bucaramanga_20260926.csv', low_memory=False)
    df_buca_clean = pd.DataFrame()
    df_buca_clean['id_siniestro'] = df_buca['ORDEN'].astype(str)
    df_buca_clean['ciudad'] = 'Bucaramanga'
    df_buca_clean['fecha_hora'] = pd.to_datetime(df_buca['FECHA'], errors='coerce')
    df_buca_clean['anio'] = df_buca['AÑO']
    df_buca_clean['mes'] = df_buca['MES']
    df_buca_clean['dia_semana'] = df_buca['DÍA']
    df_buca_clean['hora'] = pd.to_datetime(df_buca['HORA'], format='%Y-%m-%dT%H:%M:%S.%f', errors='coerce').dt.hour
    df_buca_clean['gravedad'] = df_buca['GRAVEDAD'].apply(normalizar_gravedad)
    df_buca_clean['clase_accidente'] = 'OTRO'
    df_buca_clean['direccion'] = np.nan
    df_buca_clean['barrio_localidad'] = df_buca['BARRIO'].fillna(df_buca['COMUNA'])
    df_buca_clean['latitud'] = np.nan
    df_buca_clean['longitud'] = np.nan
    dfs_procesados.append(df_buca_clean)
    print(f"   ✅ Bucaramanga: {len(df_buca_clean):,} registros")
except Exception as e:
    print(f"   ❌ Error en Bucaramanga: {e}")

# 2. BARRANQUILLA
try:
    print("🔄 Procesando Barranquilla...")
    df_baq = pd.read_csv('Accidentalidad_en_Barranquilla_20260926.csv', low_memory=False)
    df_baq_clean = pd.DataFrame()
    df_baq_clean['id_siniestro'] = [f"BAQ_{i}" for i in range(len(df_baq))]
    df_baq_clean['ciudad'] = 'Barranquilla'
    df_baq_clean['fecha_hora'] = pd.to_datetime(df_baq['FECHA_ACCIDENTE'], errors='coerce')
    df_baq_clean['anio'] = df_baq['AÑO_ACCIDENTE']
    df_baq_clean['mes'] = df_baq['MES_ACCIDENTE']
    df_baq_clean['dia_semana'] = df_baq['DIA_ACCIDENTE']
    df_baq_clean['hora'] = pd.to_datetime(df_baq['HORA_ACCIDENTE'], format='%I:%M:%S:%p', errors='coerce').dt.hour
    df_baq_clean['gravedad'] = df_baq['GRAVEDAD_ACCIDENTE'].apply(normalizar_gravedad)
    df_baq_clean['clase_accidente'] = df_baq['CLASE_ACCIDENTE'].apply(normalizar_clase)
    df_baq_clean['direccion'] = df_baq['SITIO_EXACTO_ACCIDENTE']
    df_baq_clean['barrio_localidad'] = np.nan
    df_baq_clean['latitud'] = np.nan
    df_baq_clean['longitud'] = np.nan
    dfs_procesados.append(df_baq_clean)
    print(f"   ✅ Barranquilla: {len(df_baq_clean):,} registros")
except Exception as e:
    print(f"   ❌ Error en Barranquilla: {e}")

# 3. ENVIGADO
try:
    print("🔄 Procesando Envigado...")
    df_env = pd.read_csv('Accidentalidad_Municipio_de__Envigado_20260926.csv', low_memory=False)
    df_env_clean = pd.DataFrame()
    df_env_clean['id_siniestro'] = df_env['RADICADO'].astype(str)
    df_env_clean['ciudad'] = 'Envigado'
    df_env_clean['fecha_hora'] = pd.to_datetime(df_env['FECHA'], errors='coerce')
    df_env_clean['anio'] = df_env_clean['fecha_hora'].dt.year
    df_env_clean['mes'] = df_env_clean['fecha_hora'].dt.month
    df_env_clean['dia_semana'] = df_env['DÍA DE LA SEMANA']
    df_env_clean['hora'] = pd.to_datetime(df_env['HORA'], format='%I:%M:%S %p', errors='coerce').dt.hour
    df_env_clean['gravedad'] = df_env['GRAVEDAD'].apply(normalizar_gravedad)
    df_env_clean['clase_accidente'] = df_env['CLASE DE ACCIDENTE'].apply(normalizar_clase)
    df_env_clean['direccion'] = df_env['DIRECCIÓN']
    df_env_clean['barrio_localidad'] = df_env['BARRIO']
    df_env_clean['latitud'] = np.nan
    df_env_clean['longitud'] = np.nan
    dfs_procesados.append(df_env_clean)
    print(f"   ✅ Envigado: {len(df_env_clean):,} registros")
except Exception as e:
    print(f"   ❌ Error en Envigado: {e}")

# 4. BOGOTÁ (Histórico CSV)
try:
    print("🔄 Procesando Bogotá (Histórico CSV)...")
    df_bog = pd.read_csv('historico_siniestros_bogota_d.c_-.csv', low_memory=False)
    df_bog_clean = pd.DataFrame()
    df_bog_clean['id_siniestro'] = df_bog['FORMULARIO'].fillna(df_bog['OBJECTID']).astype(str)
    df_bog_clean['ciudad'] = 'Bogotá'
    df_bog_clean['fecha_hora'] = pd.to_datetime(df_bog['FECHA_HORA_ACC'], errors='coerce')
    df_bog_clean['anio'] = df_bog['ANO_OCURRENCIA_ACC']
    df_bog_clean['mes'] = df_bog_clean['fecha_hora'].dt.month
    df_bog_clean['dia_semana'] = df_bog_clean['fecha_hora'].dt.day_name()
    df_bog_clean['hora'] = df_bog_clean['fecha_hora'].dt.hour
    df_bog_clean['gravedad'] = df_bog['GRAVEDAD'].apply(normalizar_gravedad)
    df_bog_clean['clase_accidente'] = df_bog['CLASE_ACC'].apply(normalizar_clase)
    df_bog_clean['direccion'] = df_bog['DIRECCION']
    df_bog_clean['barrio_localidad'] = df_bog['LOCALIDAD']
    df_bog_clean['latitud'] = df_bog['LATITUD']
    df_bog_clean['longitud'] = df_bog['LONGITUD']
    dfs_procesados.append(df_bog_clean)
    print(f"   ✅ Bogotá CSV: {len(df_bog_clean):,} registros")
except Exception as e:
    print(f"   ❌ Error en Bogotá CSV: {e}")

# 5. MEDELLÍN (Incidentes Viales)
try:
    print("🔄 Procesando Medellín (Incidentes Viales)...")
    df_med = pd.read_csv('incidentes_viales.csv', low_memory=False)
    df_med_clean = pd.DataFrame()
    df_med_clean['id_siniestro'] = df_med['NRO_RADICADO'].fillna(df_med['EXPEDIENTE']).astype(str)
    df_med_clean['ciudad'] = 'Medellín'
    df_med_clean['fecha_hora'] = pd.to_datetime(df_med['FECHA_ACCIDENTES'], errors='coerce')
    df_med_clean['anio'] = df_med['AÑO']
    df_med_clean['mes'] = df_med['MES']
    df_med_clean['dia_semana'] = df_med_clean['fecha_hora'].dt.day_name()
    df_med_clean['hora'] = df_med_clean['fecha_hora'].dt.hour
    df_med_clean['gravedad'] = df_med['GRAVEDAD_ACCIDENTE'].apply(normalizar_gravedad)
    df_med_clean['clase_accidente'] = df_med['CLASE_ACCIDENTE'].apply(normalizar_clase)
    df_med_clean['direccion'] = df_med['DIRECCION']
    df_med_clean['barrio_localidad'] = df_med['BARRIO'].fillna(df_med['COMUNA'])
    df_med_clean['latitud'] = df_med['Y']
    df_med_clean['longitud'] = df_med['X']
    dfs_procesados.append(df_med_clean)
    print(f"   ✅ Medellín: {len(df_med_clean):,} registros")
except Exception as e:
    print(f"   ❌ Error en Medellín: {e}")

# 6. SINIESTRALIDAD (2016-2024 CSV)
try:
    print("🔄 Procesando Siniestralidad (2016-2024)...")
    # Removido low_memory para compatibilidad con engine='python'
    df_sin = pd.read_csv('siniestralidad_2016_2024.csv', sep=None, engine='python', encoding='latin-1')
    df_sin_clean = pd.DataFrame()
    df_sin_clean['id_siniestro'] = df_sin['Código'].astype(str)
    df_sin_clean['ciudad'] = 'Nacional / Indefinido'
    df_sin_clean['fecha_hora'] = pd.to_datetime(df_sin['Fecha'], errors='coerce')
    df_sin_clean['anio'] = df_sin_clean['fecha_hora'].dt.year
    df_sin_clean['mes'] = df_sin_clean['fecha_hora'].dt.month
    df_sin_clean['dia_semana'] = df_sin_clean['fecha_hora'].dt.day_name()
    df_sin_clean['hora'] = df_sin_clean['fecha_hora'].dt.hour
    df_sin_clean['gravedad'] = df_sin['Tipo confirmado'].apply(normalizar_gravedad)
    df_sin_clean['clase_accidente'] = df_sin['Tipo clase de accidente'].apply(normalizar_clase)
    df_sin_clean['direccion'] = df_sin['Dirección reporte']
    df_sin_clean['barrio_localidad'] = np.nan
    df_sin_clean['latitud'] = np.nan
    df_sin_clean['longitud'] = np.nan
    dfs_procesados.append(df_sin_clean)
    print(f"   ✅ Siniestralidad 2016-2024: {len(df_sin_clean):,} registros")
except Exception as e:
    print(f"   ❌ Error en Siniestralidad 2016-2024: {e}")

# 7. BOGOTÁ (Consolidado Excel)
try:
    print("🔄 Procesando Bogotá Consolidado (Excel)...")
    df_bog_xls = pd.read_excel('siniestros_viales_consolidados_bogota_dc.xlsx')
    df_bog_xls_clean = pd.DataFrame()
    df_bog_xls_clean['id_siniestro'] = [f"BOG_XLS_{i}" for i in range(len(df_bog_xls))]
    df_bog_xls_clean['ciudad'] = 'Bogotá'
    
    col_fecha = [c for c in df_bog_xls.columns if 'FECHA' in c.upper() or 'DATE' in c.upper()]
    fecha_col = col_fecha[0] if col_fecha else df_bog_xls.columns[0]
    
    df_bog_xls_clean['fecha_hora'] = pd.to_datetime(df_bog_xls[fecha_col], errors='coerce')
    df_bog_xls_clean['anio'] = df_bog_xls_clean['fecha_hora'].dt.year
    df_bog_xls_clean['mes'] = df_bog_xls_clean['fecha_hora'].dt.month
    df_bog_xls_clean['dia_semana'] = df_bog_xls_clean['fecha_hora'].dt.day_name()
    df_bog_xls_clean['hora'] = df_bog_xls_clean['fecha_hora'].dt.hour
    
    col_grav = [c for c in df_bog_xls.columns if 'GRAVEDAD' in c.upper()]
    df_bog_xls_clean['gravedad'] = df_bog_xls[col_grav[0]].apply(normalizar_gravedad) if col_grav else 'OTRO'
    
    col_clase = [c for c in df_bog_xls.columns if 'CLASE' in c.upper()]
    df_bog_xls_clean['clase_accidente'] = df_bog_xls[col_clase[0]].apply(normalizar_clase) if col_clase else 'OTRO'
    
    df_bog_xls_clean['direccion'] = np.nan
    df_bog_xls_clean['barrio_localidad'] = np.nan
    df_bog_xls_clean['latitud'] = np.nan
    df_bog_xls_clean['longitud'] = np.nan
    dfs_procesados.append(df_bog_xls_clean)
    print(f"   ✅ Bogotá Excel: {len(df_bog_xls_clean):,} registros")
except Exception as e:
    print(f"   ❌ Error en Bogotá Excel: {e}")

# CONSOLIDADO FINAL
df_consolidado = pd.concat(dfs_procesados, ignore_index=True)

print("\n" + "="*80)
print(f"🎉 CONSOLIDACIÓN COMPLETADA EXITOSAMENTE 🎉")
print(f"📊 Total de registros finales: {len(df_consolidado):,}")
print("="*80)
```

```python
import pymongo
from pymongo import MongoClient
import pandas as pd
import numpy as np

print("⏳ Conectando a MongoDB e iniciando la limpieza y carga masiva...")

client = MongoClient("mongodb://localhost:27017/")
db = client["db_siniestralidad_vial"]
collection = db["siniestros"]

# Limpiar colección si ya existía para evitar duplicados
collection.drop()

df_mongo = df_consolidado.copy()

# ==========================================
# 🧹 LIMPIEZA FINAL DE CALIDAD DE DATOS (QA)
# ==========================================

# 1. Limpiar Años (quitar saltos de línea "\r", decimales y convertir a entero puro)
df_mongo['anio'] = pd.to_numeric(df_mongo['anio'].astype(str).str.replace(r'\D', '', regex=True), errors='coerce')
df_mongo['anio'] = df_mongo['anio'].fillna(0).astype(int) # 0 para los no reportados

# 2. Unificar Días de la Semana (Inglés a Español y todo en mayúsculas)
dias_map = {
    'MONDAY': 'LUNES', 'TUESDAY': 'MARTES', 'WEDNESDAY': 'MIÉRCOLES',
    'THURSDAY': 'JUEVES', 'FRIDAY': 'VIERNES', 'SATURDAY': 'SÁBADO', 'SUNDAY': 'DOMINGO'
}
df_mongo['dia_semana'] = df_mongo['dia_semana'].astype(str).str.upper().str.strip()
df_mongo['dia_semana'] = df_mongo['dia_semana'].replace(dias_map)
df_mongo['dia_semana'] = df_mongo['dia_semana'].replace(['NAN', 'NAT', 'NONE', ''], 'NO REPORTADO')

# 3. Formatear fechas a String para compatibilidad BSON/JSON
df_mongo['fecha_hora'] = df_mongo['fecha_hora'].astype(str)

# ==========================================
# 📥 CARGA A MONGODB
# ==========================================
records = df_mongo.to_dict('records')
batch_size = 50000
total_records = len(records)

for i in range(0, total_records, batch_size):
    lote = records[i:i + batch_size]
    collection.insert_many(lote)
    print(f"   📥 Insertados {min(i + batch_size, total_records):,} de {total_records:,} registros...")

print("\n" + "="*80)
print("✅ ¡LIMPIEZA Y CARGA A MONGODB COMPLETADA CON ÉXITO!")
print(f"📌 Total de documentos guardados en MongoDB: {collection.count_documents({}):,}")
print("="*80)
```
